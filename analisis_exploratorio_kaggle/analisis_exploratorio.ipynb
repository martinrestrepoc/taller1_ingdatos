# Análisis exploratorio del rendimiento de jugadores - FIFA World Cup 2026

## 1. Importación de librerías

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## 2. Carga y selección de columnas

```python
columnas_seleccionadas = [
    'player_id',
    'match_id',
    'match_date',
    'team',
    'opponent_team',
    'position',
    'preferred_foot',
    'tournament_stage',
    'match_result',
    'age',
    'market_value_eur',
    'minutes_played',
    'goals',
    'assists',
    'expected_goals_xg',
    'pass_accuracy',
    'distance_covered_km',
    'player_rating'
]
```

```python
data = pd.read_csv(
    'fifa_world_cup_2026_player_performance.csv',
    usecols=columnas_seleccionadas
)
data.head()
```

## 5. Clasificación de las variables

```python
columns_date = ['match_date']

columns_categorical = [
    'team',
    'opponent_team',
    'position',
    'preferred_foot',
    'tournament_stage',
    'match_result'
]

columns_numerical = [
    'age',
    'market_value_eur',
    'minutes_played',
    'goals',
    'assists',
    'expected_goals_xg',
    'pass_accuracy',
    'distance_covered_km',
    'player_rating'
]
```

```python
len(columns_date) + len(columns_categorical) + len(columns_numerical) == data.shape[1] # Verifica que todas las columnas de análisis estén clasificadas.
```

## 6. Conversión de tipos de datos

```python
for col_date in columns_date:
    data[col_date] = pd.to_datetime(data[col_date]) # Convierte las columnas de fecha a tipo datetime.

for col_categorical in columns_categorical:
    data[col_categorical] = data[col_categorical].astype('category') # Convierte las columnas categóricas a tipo category.
```

# Análisis Univariante

## 7. Análisis univariante categórico

```python
for col in columns_categorical:
    print(f"\n{'='*70}")
    print(f"ANÁLISIS DE: {col.upper()}")
    print(f"{'='*70}")

    # 1. Cantidad de categorías
    num_categories = data[col].nunique()
    print(f"\nCantidad de categorías: {num_categories}")

    # 2. Tabla de conteo y porcentajes
    value_counts = data[col].value_counts()
    percentage = (value_counts / len(data) * 100).round(2)

    table = pd.DataFrame({
        'Categoría': value_counts.index,
        'Conteo': value_counts.values,
        'Porcentaje (%)': percentage.values
    })
    table = table.reset_index(drop=True)
    print("\nTabla de conteo y porcentajes:")
    display(table)

    # 3. Gráficos para variables con menos de 10 categorías
    if num_categories < 10:
        fig, axes = plt.subplots(1, 2, figsize=(14, 5))

        colors_array = plt.cm.Set2(range(num_categories))

        sns.countplot(
            data=data,
            x=col,
            ax=axes[0],
            order=value_counts.index,
            legend=False,
            width=0.75
        )

        for i, patch in enumerate(axes[0].patches):
            patch.set_facecolor(colors_array[i])

        axes[0].set_title(f'Conteo de {col}', fontsize=12, fontweight='bold')
        axes[0].set_xlabel(col, fontsize=10)
        axes[0].set_ylabel('Conteo', fontsize=10)
        axes[0].tick_params(axis='x', rotation=45)

        axes[1].pie(
            value_counts.values,
            labels=value_counts.index,
            autopct='%1.3f%%',
            startangle=180,
            colors=colors_array
        )
        axes[1].set_title(f'Distribución de {col}', fontsize=12, fontweight='bold')

        plt.tight_layout()
        plt.show()
    else:
        print(f"\nNota: {col} tiene {num_categories} categorías (≥10), no se generó gráfico.")
```

## 8. Análisis univariante numérico

```python
for col in columns_numerical:
    print(f"\n{'='*70}")
    print(f"ANÁLISIS DE: {col.upper()}")
    print(f"{'='*70}")

    # Tabla de estadística descriptiva
    stats = data[col].describe().to_frame().T
    print("\nEstadística descriptiva:")
    display(stats)

    # Histograma arriba y boxplot abajo
    fig, axes = plt.subplots(2, 1, figsize=(10, 8))

    sns.histplot(data=data, x=col, ax=axes[0], color='steelblue')
    axes[0].set_title(f'Histograma de {col}', fontsize=12, fontweight='bold')
    axes[0].set_xlabel(col, fontsize=10)
    axes[0].set_ylabel('Frecuencia', fontsize=10)

    sns.boxplot(data=data, x=col, ax=axes[1], color='steelblue')
    axes[1].set_title(f'Boxplot de {col}', fontsize=12, fontweight='bold')
    axes[1].set_xlabel(col, fontsize=10)

    plt.tight_layout()
    plt.show()
```

## 9. Análisis univariante temporal

Los conteos de esta sección representan registros jugador-partido. No representan directamente la cantidad de partidos disputados en cada fecha.

```python
for col in columns_date:
    print(f"\n{'='*70}")
    print(f"ANÁLISIS DE: {col.upper()}")
    print(f"{'='*70}")

    # Fecha mínima y máxima
    print(f"\nFecha mínima: {data[col].min()}")
    print(f"Fecha máxima: {data[col].max()}")

    # Fecha con más registros jugador-partido
    registros_por_fecha = data[col].value_counts()
    print(
        f"Fecha con más registros jugador-partido: {registros_por_fecha.index[0]} "
        f"({registros_por_fecha.iloc[0]} registros)"
    )

    fig, axes = plt.subplots(3, 1, figsize=(12, 12))

    # Distribución por año
    sns.countplot(x=data[col].dt.year, ax=axes[0], color='steelblue')
    axes[0].set_title('Distribución por año', fontsize=12, fontweight='bold')
    axes[0].set_xlabel('Año', fontsize=10)
    axes[0].set_ylabel('Cantidad de registros', fontsize=10)

    # Distribución por mes
    meses = [
        'Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio',
        'Julio', 'Agosto', 'Septiembre', 'Octubre', 'Noviembre', 'Diciembre'
    ]
    sns.countplot(x=data[col].dt.month, order=range(1, 13), ax=axes[1], color='steelblue')
    axes[1].set_title('Distribución por mes', fontsize=12, fontweight='bold')
    axes[1].set_xlabel('Mes', fontsize=10)
    axes[1].set_ylabel('Cantidad de registros', fontsize=10)
    axes[1].set_xticks(range(12))
    axes[1].set_xticklabels(meses, rotation=45)

    # Distribución por día de la semana
    dias = ['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo']
    sns.countplot(x=data[col].dt.dayofweek, order=range(7), ax=axes[2], color='steelblue')
    axes[2].set_title('Distribución por día de la semana', fontsize=12, fontweight='bold')
    axes[2].set_xlabel('Día de la semana', fontsize=10)
    axes[2].set_ylabel('Cantidad de registros', fontsize=10)
    axes[2].set_xticks(range(7))
    axes[2].set_xticklabels(dias)

    plt.tight_layout()
    plt.show()
```

## 11. Análisis bivariante categórico

```python
# Define aquí las dos columnas que quieres cruzar
columna_filas = 'tournament_stage'
columna_columnas = 'match_result'

# Crosstab con conteos absolutos
crosstab = pd.crosstab(
    data[columna_filas],
    data[columna_columnas]
).astype(int)

valor_min = crosstab.values.min()
valor_max = crosstab.values.max()

plt.figure(figsize=(12, 8))
sns.heatmap(
    crosstab,
    annot=True,
    fmt='d',
    cmap='RdBu_r',
    vmin=valor_min,
    vmax=valor_max,
    linewidths=2,
    linecolor='white',
    cbar_kws={'shrink': 0.85}
)
plt.title(f'Crosstab entre {columna_filas} y {columna_columnas}', fontsize=14, fontweight='bold')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Define aquí las dos columnas que quieres cruzar
columna_filas = 'position'
columna_columnas = 'match_result'

plt.figure(figsize=(12, 8))
sns.countplot(
    data=data,
    x=columna_filas,
    hue=columna_columnas,
    order=data[columna_filas].value_counts().index,
    hue_order=data[columna_columnas].value_counts().index,
    palette='tab20'
)
plt.title(f'Diagrama de barras entre {columna_filas} y {columna_columnas}', fontsize=14, fontweight='bold')
plt.xlabel(columna_filas, fontsize=10)
plt.ylabel('Conteo', fontsize=10)
plt.xticks(rotation=45, ha='right')
plt.legend(title=columna_columnas, bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## 12. Análisis categórico + numérico

```python
# Define aquí una variable categórica y una numérica
columna_categorica = 'position'
columna_numerica = 'player_rating'

print(f'Variable categórica: {columna_categorica}')
print(f'Variable numérica: {columna_numerica}')

# Boxplot comparativo por categoría
plt.figure(figsize=(12, 6))
sns.boxplot(
    data=data,
    x=columna_categorica,
    y=columna_numerica,
    color='steelblue'
)
plt.title(f'Boxplot de {columna_numerica} por {columna_categorica}', fontsize=14, fontweight='bold')
plt.xlabel(columna_categorica, fontsize=10)
plt.ylabel(columna_numerica, fontsize=10)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Histogramas en subplots separados por categoría
categorias = data[columna_categorica].dropna().unique()
num_categorias = len(categorias)
num_columnas = 2
num_filas = (num_categorias + num_columnas - 1) // num_columnas

fig, axes = plt.subplots(num_filas, num_columnas, figsize=(14, 4 * num_filas))
axes = axes.flatten() if num_categorias > 1 else [axes]

for i, categoria in enumerate(categorias):
    subset = data[data[columna_categorica] == categoria]
    sns.histplot(
        data=subset,
        x=columna_numerica,
        kde=True,
        color='steelblue',
        ax=axes[i]
    )
    axes[i].set_title(f'{columna_categorica} = {categoria}', fontsize=12, fontweight='bold')
    axes[i].set_xlabel(columna_numerica, fontsize=10)
    axes[i].set_ylabel('Frecuencia', fontsize=10)

for j in range(num_categorias, len(axes)):
    axes[j].axis('off')

fig.suptitle(f'Histogramas de {columna_numerica} por {columna_categorica}', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 13. Análisis temporal + categórico/numérico

```python
# Define aquí una variable numérica y una categórica para analizar su evolución temporal
columna_numerica = 'player_rating'
columna_categorica = 'match_result'
columna_tiempo = 'match_date'

print(f'Variable numérica: {columna_numerica}')
print(f'Variable categórica: {columna_categorica}')
print(f'Variable temporal: {columna_tiempo}')

# Evolución temporal de la calificación promedio por fecha
serie_numerica = data.groupby(columna_tiempo)[columna_numerica].mean().reset_index()

plt.figure(figsize=(12, 5))
sns.lineplot(
    data=serie_numerica,
    x=columna_tiempo,
    y=columna_numerica,
    color='steelblue'
)
plt.title(f'Evolución temporal del promedio de {columna_numerica}', fontsize=14, fontweight='bold')
plt.xlabel(columna_tiempo, fontsize=10)
plt.ylabel(f'Promedio de {columna_numerica}', fontsize=10)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Evolución temporal del conteo de registros por resultado
serie_categorica = (
    data.groupby([columna_tiempo, columna_categorica])
        .size()
        .reset_index(name='Conteo')
)

categorias = serie_categorica[columna_categorica].dropna().unique()
num_categorias = len(categorias)
num_columnas = 3
num_filas = (num_categorias + num_columnas - 1) // num_columnas

fig, axes = plt.subplots(
    num_filas,
    num_columnas,
    figsize=(16, 4 * num_filas),
    sharex=True,
    sharey=False
)
axes = axes.flatten() if num_categorias > 1 else [axes]

for i, categoria in enumerate(categorias):
    subset = serie_categorica[serie_categorica[columna_categorica] == categoria]
    sns.lineplot(
        data=subset,
        x=columna_tiempo,
        y='Conteo',
        ax=axes[i],
        color='steelblue'
    )
    axes[i].set_title(f'{columna_categorica} = {categoria}', fontsize=12, fontweight='bold')
    axes[i].set_xlabel(columna_tiempo, fontsize=10)
    axes[i].set_ylabel('Conteo', fontsize=10)
    axes[i].tick_params(axis='x', rotation=45)

for j in range(num_categorias, len(axes)):
    axes[j].axis('off')

fig.suptitle(f'Evolución temporal de {columna_categorica} por subgrupos', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

# Cinco preguntas de análisis

## Pregunta 1

**¿Los jugadores con mayor cantidad de goles esperados (`expected_goals_xg`) suelen marcar más goles?**

```python
correlacion_xg_goles = data[['expected_goals_xg', 'goals']].corr().round(3)
display(correlacion_xg_goles)

sns.pairplot(data[['expected_goals_xg', 'goals']])
plt.show()
```

**Conclusión:** Existe una relación positiva, pero débil, entre los goles esperados y los goles marcados. La correlación es aproximadamente **0.150**, por lo que un mayor `expected_goals_xg` se asocia con más goles, aunque por sí solo no explica gran parte de su variación.

## Pregunta 2

**¿Cómo cambia la calificación del jugador (`player_rating`) según su posición?**

```python
calificacion_por_posicion = data.groupby('position')['player_rating'].describe().round(3)
display(calificacion_por_posicion)

plt.figure(figsize=(12, 6))
sns.boxplot(
    data=data,
    x='position',
    y='player_rating',
    color='steelblue'
)
plt.title('Calificación del jugador por posición', fontsize=14, fontweight='bold')
plt.xlabel('Posición')
plt.ylabel('Calificación del jugador')
plt.tight_layout()
plt.show()
```

**Conclusión:** Los delanteros presentan el promedio más alto de calificación (**3.884**), seguidos de los mediocampistas (**3.861**) y defensores (**3.790**). Los porteros tienen un promedio considerablemente menor (**2.068**) y una mediana de **0**, por lo que la distribución de su calificación difiere de la de las demás posiciones.

## Pregunta 3

**¿Los jugadores de equipos ganadores obtienen mejores calificaciones que los de equipos que empatan o pierden?**

```python
calificacion_por_resultado = data.groupby('match_result')['player_rating'].describe().round(3)
display(calificacion_por_resultado)

plt.figure(figsize=(12, 6))
sns.boxplot(
    data=data,
    x='match_result',
    y='player_rating',
    color='steelblue'
)
plt.title('Calificación del jugador según el resultado', fontsize=14, fontweight='bold')
plt.xlabel('Resultado del partido')
plt.ylabel('Calificación del jugador')
plt.tight_layout()
plt.show()
```

**Conclusión:** Los jugadores de equipos ganadores tienen una calificación media ligeramente superior (**3.685**) a la de quienes empatan (**3.605**) o pierden (**3.606**). La diferencia es pequeña, aunque la mediana también aumenta de **5.4** a **5.6** en las victorias.

## Pregunta 4

**¿La distancia recorrida por los jugadores cambia según su posición?**

```python
distancia_por_posicion = data.groupby('position')['distance_covered_km'].describe().round(3)
display(distancia_por_posicion)

plt.figure(figsize=(12, 6))
sns.boxplot(
    data=data,
    x='position',
    y='distance_covered_km',
    color='steelblue'
)
plt.title('Distancia recorrida según la posición', fontsize=14, fontweight='bold')
plt.xlabel('Posición')
plt.ylabel('Distancia recorrida (km)')
plt.tight_layout()
plt.show()
```

**Conclusión:** Los mediocampistas recorren la mayor distancia promedio (**4.499 km**), seguidos de los delanteros (**4.125 km**) y defensores (**3.993 km**). Los porteros presentan la menor distancia promedio (**2.411 km**), lo cual evidencia diferencias de desplazamiento según la posición.

## Pregunta 5

**¿Cómo evoluciona la calificación promedio de los jugadores durante el torneo?**

```python
calificacion_por_fecha = (
    data.groupby('match_date')['player_rating']
        .mean()
        .reset_index()
)
display(calificacion_por_fecha)

plt.figure(figsize=(12, 5))
sns.lineplot(
    data=calificacion_por_fecha,
    x='match_date',
    y='player_rating',
    color='steelblue'
)
plt.title('Evolución de la calificación promedio durante el torneo', fontsize=14, fontweight='bold')
plt.xlabel('Fecha del partido')
plt.ylabel('Calificación promedio')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

**Conclusión:** La calificación promedio permanece relativamente estable durante el torneo, con valores diarios entre aproximadamente **3.544** y **3.701**. Aunque pasa de **3.576** en la primera fecha a **3.646** en la última, las variaciones son pequeñas y no se observa una tendencia fuerte de aumento o disminución.
